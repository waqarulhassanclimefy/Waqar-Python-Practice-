```python
# A Climefy Activity String
activity = "Diesel generator combustion"

# Length of the string
print(len(activity))

# First character (position 0)
print(activity[0])

# Fourth character (position 3)
print(activity[3])

# Last character (negative index)
print(activity[-1])

# Third to last character
print(activity[-3])
```

```python
activity = "Diesel generator combustion"

# First 6 character (position 0 to 5)
print(activity[0:6])

# Shortcut
print(activity[:6])

# From position 7 to the end
print(activity[7:])

# Last 10 Character
print(activity[-10:])

# Every second character 
print(activity[::2])

# Reverse the whole string
print(activity[::-1])
```

```python
# The same activity, entered different ways by different clients
entry1 = "diesel"
entry2 = "Diesel"
entry3 = "DIESEL"
entry4 = "dIeSel"

print(entry1 == entry2)

print(entry1 == entry2.lower())
print(entry3.lower() == entry4.lower())

print(entry1.upper())
print(entry3.lower())
print("natural gas" .title())
print("Scope 1 emissions" .capitalize())
```

```python
# Messy input from client's CSV file
raw = " Natural Gas "
print(f"Original length: {len(raw)}")

# strip whitespace from both ends
cleaned = raw.strip()
print(f"After strip: '{cleaned}' (length {len(cleaned)})")

# Strip only left / only right 
print(f"left only: '{raw.lstrip()}'")
print(f"right only: '{raw.rstrip()}'")

# Replace inconsistent separators
messy = "natural-gas_combustion"
print(messy.replace("-"," ").replace("_", " "))
```

```python
# Splitting a combined field 
combined = "F001-Diesel-Scope1"
A1 = (combined.split("-"))
print(A1)

# Now we have separate fields we can use
facility_id = A1[0]
activity = A1[1]
scope = A1[2]
print(f"Facility: {facility_id}, Activity: {activity}, Scope: {scope}")

```

```python
# Joining a list back into a string
B1 = "_".join(A1)
print(B1)

# Find where a substring starts
text = "Scope 3 emissions catagory 15"
print(text.find("Scope"))
print(text.find("15"))
print(text.find("Scopes"))

# Count Occurrences
print(text.count("1"))

```
