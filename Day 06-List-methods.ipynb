```python
activities = ["diesel", "natural gas", "petrol"]

activities.append("coal")
activities.insert(2, "refrigrents")
activities.extend(["biomass", "hydrogen"])
print(activities)

# The classic bug - what does .append return?
result = activities.append("waste")
print("Return value:", result)

# .count - tally a value
flags = [True, True, False, True, False, True, False]
print("Verified:", flags.count(True))
print("Pending:", flags.count(False))
```

```python
# note: F002 appear twice
facilities = ["F001", "F002", "F003", "F004", "F002"]

facilities.remove("F002")
print("After Remove:", facilities)

last = facilities.pop()
first = facilities.pop(0)
print("Popped:", first, "and", last)
print("Now:", facilities)

# .index with start argument - Find duplicates
nums = [10, 20, 30, 20, 40, 20]
first_20 = nums.index(20)
next_20 = nums.index(20, first_20 + 1)
last_20 = nums.index(20, next_20 + 1)
print("20s at:", first_20, ",", next_20, "and", last_20)

# Safe pattern - check before .remove/.index
if "F002" in facilities:
    facilities.remove("F002")
else:
    print("F002 not present - Skipped")
```

```python
monthly = [45.2, 41.8, 52.7, 52.7, 38.9, 47.3, 55.1]

# sorted
print("Desc order:", sorted(monthly, reverse=True))
print("Original oder:", monthly)

#.sort
monthly.sort()
print("simple sort:", monthly)

# The classic bug
a = [3, 1, 2]
b = a.sort()
c = sorted(a)
print("b:", b, "c:", c)

# key = - sort by somthing derived
fuels = ["diesel", "coal", "natural gas", "petrol"]
print("by lenght:", sorted(fuels, key=len))
```
