```python
factors = {"diesel": 2.68, "gas": 2.02, "petrol": 2.31}

# Each method returns a view - Not a plain list
print("keys type:", type(factors.keys()))
print("keys:", factors.keys())
print("values:", factors.values())
print("items:", factors.items())

# Wrap in lists() to get a sanpshot as a noral list
key_list = list(factors.keys())
print("As list:", key_list)
print("First key:", key_list[0])

# Views are LIVE - they reflect later changes
view = factors.keys()
factors["coal"] = 2.42
print("view:", view) 

# Aggregate one liners over values
print("Sum:", sum(factors.values()))
print("Max:", max(factors.values()))
print("2.68 in values:", 2.18 in factors.values())

```

```python
tonnes = {"jan": 45.2, "Feb": 41.8, "Mar": 52.7}

# 1. Defualt loop - Gives you the keys
print("--- keys ---")
for z in tonnes:
    print(f" {z}: {tonnes[z]}")

# 2 .items()- the key and value together
print("--- items ---")
for month, val in tonnes.items():
    print(f" {month}: {val} tCO2e")

# 3 Sort alphebetically by key
print("--- sorted by key ---")
for m in sorted(tonnes):
    print(f" {m} -> {tonnes[m]}")

# 4 Sort by value, highest first
ranked = sorted(tonnes.items(),
                key=lambda x: x[0],
                reverse=True)
print("--- ranked by value ---")
for month, val in ranked:
    print(f" {month}: {val}")

# 5 Enumerate + items - numbered report
print("--- numbered ---")
for i, (m, v) in enumerate(tonnes.items(), start=1):
    print(f" {i}, {m}: {v}")
```

```python
# Acess - double bracket [outer] [inner]

# Facility ID -> its record dict
facilities = {
    "F001": {"fuel": "diesel", "tonnes":45.2},
    "F002": {"fuel": "gas", "tonnes": 38.7},
    "F003": {"fuel": "petrol", "tonnes": 22.4},
}
print("F002 tonnes:", facilities["F002"]["tonnes"])
print("F002 fuel:", facilities["F002"]["fuel"])

# Add new inner key to an exsisting record
facilities["F001"]["scope"] = "Scope 1"
print("F001 now:", facilities["F001"])

# Add a whole new facility 
facilities["F004"] = {"fuel": "coal", "tonnes": 33.1}
print("added F004")

# Iterate - outer layer + full inner record
for fid, record in facilities.items():
    fuel = record.get("fuel", "?")
    t = record.get("tonnes", "")
    print(f" {fid}: {t} tCO2e ({fuel})")

# Chained safe access - never crashes
val = facilities.get("F004", {}).get("tonnes", 0)
print(f"Checking F004 tonnes: {val}")
```

```python
# Every row is a dict - Same keys across rows
records = [
    {"id":"F001", "fuel":"diesel", "tonnes":45.2},
    {"id":"F002", "fuel":"gas", "tonnes":38.7},
    {"id":"F003", "fuel":"petrol", "tonnes":22.4},
    {"id":"F004", "fuel":"diesel", "tonnes":51.9},
]

# Access one Row by position, then a field
print("Row 1 fuel:", records[1]["fuel"])

# Extract a column using a comprehension
tonnes_col = [j["tonnes"] for j in records]
fuel_col = [p["fuel"] for p in records]
print("tonnes col:", tonnes_col)
print("fuel col:", fuel_col)

# Filter Rows - Records above 30 tonnes
high = [r for r in records if r["tonnes"] > 30]
print("high rows:", len(high))

# *** The Preview = this list is a dataframe
import pandas as pd
df = pd.DataFrame(records)
print(df)
print("type:", type(df).__name__)
```

```python
# Group/catagorizing - the setdefualt trick
# Group facilities by their scope
records = [
    {"id":"F001", "Scope":"1", "tonnes":45.2},
    {"id":"F002", "Scope":"2", "tonnes":38.7},
    {"id":"F003", "Scope":"1", "tonnes":22.4},
    {"id":"F004", "Scope":"2", "tonnes":51.9},
    {"id":"F005", "Scope":"3", "tonnes":33.1},
]

# .setdefault(key, []) - get or create empty list
by_scope = {}
for r in records:
    scope = r["Scope"]
    by_scope.setdefault(scope, []).append(r)

for scope, items in by_scope.items():
    print(f"scope {scope}: {len(items)} facilities")

# Compute total tonnes per group in one expression
totals = {s: sum(r["tonnes"] for r in items)
          for s, items in by_scope.items()}
print("Totals:", totals)

# Ranked the groups by total, higest first
ranked = sorted(totals.items(),
                key=lambda x: x[1], reverse=True)
print("Ranked:", ranked)


    
```

```python
# Nested Agrregation
records = [
    {"scope": "1", "fuels":"diesel", "tonnes":45.2},
    {"scope": "1", "fuels":"gas", "tonnes":38.7},
    {"scope": "2", "fuels":"grid", "tonnes":22.4},
    {"scope": "1", "fuels":"diesel", "tonnes":51.9},
    {"scope": "2", "fuels":"grid", "tonnes":29.8},
    {"scope": "1", "fuels": "gas", "tonnes":11.3},
]

# Nested totals: totals[scope][fuel], r["tonnes"]
totals = {}
for r in records:
    s, f, t = r["scope"], r["fuels"], r["tonnes"]
    totals.setdefault(s, {}).setdefault(f, 0)
    totals[s][f] += t

# print the nested breakdown:
print("Breakdown by scope & fuels")
for scope in sorted (totals):
    print(f"scope {scope}:")
    for fuel, tot in totals[scope].items():
        print(f" {fuel: <6}: {round(tot, 2)} tCO2e")

    
    
     
    
```
