```python
tonnes = 1000

# Catagorise this facility's emissions
if tonnes < 20:
    catagory = "Low"
elif tonnes < 40:
    catagory = "Medium"
elif tonnes < 100:
    catagory = "High"

print(f"{tonnes} tCO2e -> {catagory}")

# Only one branch runs - Python stops at first true
# change the tonnes value to 15,55,150 and remove else function - nothing will happen if condition is false

verified = False
if not verified:
    print(" ! record needs review")

# with else - Always one action taken
score = 85
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "Below C"
print(f"score {score} -> grade {grade}")
```

```python
a, b = 45.2, 38.7

# The six comparison operators
print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)

# Chained comparisions - Read like math!
tonnes = 45
if 20 <= tonnes < 100:
    print("In normal range")
    
# String comparison - alphabatically order
print("diesel < gas:", "ziesel" < "gas")

# "in" as a condition - membership check
approved = ["diesel", "gas", "petrol"]
fuel = "coal"
if fuel not in approved:
    print(f"not approved {fuel} not in approved list")

# "is none" - the pythonic none check
val = None
print("val is None:", val is None)
```

```python
record = {
    "id": "F001", "fuel": "diesel", "tonnes": 45.2, "verified": True,
}

# AND - Both coditions must be true
if record["tonnes"] > 20 and record["verified"]:
    print("OK valid & verified")

# OR - at least one must be true
if record["fuel"] in ("diesel", "gas"):
    print("Fossil fuel record")

# NOT - flip true to false
if not record["verified"]:
    print("Needs Varification")
else:
    print("Already Verified")

# Combining - Parens for clarity
if (record["tonnes"] > 100
    and record["verified"]
    and record["fuel"] != "unknown"):
    print("Flag for audit")

# Short circuit - "or" for dafault values
fuel_type = record.get("fuel") or "unknown"
print("fuel:", fuel_type)

# Short Circuit protects the second half
safe_rec = None
if safe_rec and safe_rec["tonnes"] > 0:
    print("processed")
else:
    print("skipped (safe_rec is None)")
    
          
```

```python
# Empty containers are FALSY
print(bool([]))
print(bool({}))
print(bool(""))
print(bool(0))
print(bool(None))

# Non-empty containers are truthy
print(bool([0]))
print(bool("hi"))

# The pythonic idiom - no need for len(x) > 0
record = []
if not record:
    print("no record to process")

records = [{"id":"F001", "AB":"F002"},
          {"id":"F001", "AB":"F002"}
          ]
if records:
    print(f"processing {len(records)} record(s)")

# "or" for defualts - Very useful pattern
x = {}
region = x.get("region") or "global"
print(f"Region: {region}")

# O is Falsy
tonnes = 0
value = tonnes or "no data"
print(f"value: {value}")

# Fix: check "is None" explicity
value = tonnes if tonnes is not None else "no data"
print(f"value: {value}")
```

```python
# Ternary expression - one line if/else
tonnes = 45.2

# VERBOSE - four lines for a simple choice 
if tonnes > 50:
    label = "High"
else:
    label = "Normal"
print("verbose:", label)

# Ternary - Same logic in one line
label = "High" if tonnes > 50 else "Normal"
print("Ternary:", label)

# Ternary inside a comprehnsion
tonnes_list = [22.4, 45.2, 68.9, 15.1, 88.3]
tags = ["High" if r > 50 else "Normal" for r in tonnes_list]
print("tags:", tags)

# Ternary with dict access 
record = {"fuel": "diesel"}
fuel = record["fuel"] if "fuel" in record else "unknown"
print("Fuel:", fuel)
```
