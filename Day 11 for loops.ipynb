```python
# Monthly diesel usage of Indus textiles - Monthwise (Jan-Dec)
diesel_1 = [12000, 11500, 12300, 13100, 12800, 13500, 14200, 14800, 13900, 13200, 12700, 12400]

# Basic for loop - accumulate the total
total = 0
for amount in diesel_1:
    total += amount

print(f"Total liters : {total}")

# Commute emission insdie the loop

```

```python
Factor = 2.68 / 1000 # tCO2e per litre
tonnes = 0
for amount in diesel_1:
    tonnes += amount * Factor

print(f"Total tCO2e : {round(tonnes, 2)}")
```

```python
# Same thing with sum() - for the one liner shortcut
check = sum(diesel_1) * Factor
print(f"check via sum: {round(check, 2)}")
```

```python
# Iterate a string - one character at a time
Scope = "S1"
for ch in Scope:
    print(ch)

# Extract initials from a facility name
name = "Karachi Head Office"
initials = ""
for ch in name:
    if ch.isupper():
        initials += ch
print(f"{name} -> {initials}")

# iterate a dict - defualt is keys
factor = {"diesel": 2.68, "gas": 2.02, "petrol": 2.31}
print("---keys---")
for fuel in factor:
    print(fuel)

print("---values---")
for v in factor.values():
    print(v)

print("---items---")
for fuel, v in factor.items():
    print(f" {fuel:<8} -> {v} kg/L")
```

```python
# range(stop) - 0, 1, 2 ....,  stop-1
for i in range(5):
    print(f"step {i}")

# range(start, stop) - from start to stop-1
print("---months 1...12---")
for month in range(1, 13):
    print(month, end=" ")
print()
# range (start, stop, step) - with step size
print("---years by 2---")
for year in range(2020, 2031, 2):
    print(year, end=" ")
print()

# Countdown - negative step
print("---countdown---")
for i in range(10, 0, -1):
    print(i, end=" ")
print()

# Range with index into a list (older pattern)
months = ["Jan", "Feb", "Mar", "Apr"]
for i in range(len(months)):
    print(f"Position {i}: {months[i]}")
```

```python
# Enumerate, Zip, reversed, - the main helpers
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
tonnes = [45.2, 41.8, 52.7, 38.9, 47.3, 55.1]

# Enumerate - index + value together 
for i, m in enumerate(months):
    print(f"{i}: {m}")

# Enumerate with start=1 for report numbering
print("---1-indexed---")
for i, m in enumerate(months, start=1):
    print(f"Month {i}: {m}")

# Zip - walk two lists in parallel
print("---zipped---")
for m, t in zip(months, tonnes):
    print(f"{m}: {t} tCO2e")

# Enumerate + Zip - three variables at once
print("---Ranked Report---")
paired = zip(months, tonnes)
for i, (m, t) in enumerate(paired, start=1):
    print(f"{i}. {m} -> {t}")

# reversed() - Walk backwards without modifying
print("---reversed---")
for m in reversed(months):
    print(m, end=" ")
print()

# Sorted() inside a for loop - order by value
print("---sorted by tonnes---")
ranked = sorted(zip(months, tonnes),
                key=lambda x: x[1], reverse=True)
for m, t in ranked:
    print(f" {m}: {t}")
```

```python
# Simple nesting - print every combination
scopes = ["1", "2", "3"]
fuels = ["diesel", "gas"]

for s in scopes: 
    for f in fuels:
        print(f" Scope {s} x {f}")
        print()

# Practical - build a facility x month totals grid
sites = ["Karachi", "Lahore", "Islamabad"]
months = ["Jan", "Feb", "Mar"]

# Start with zeros- nested dict scaffold
totals = {}
for site in sites:
    totals[site] = {}
    for month in months:
        totals[site][month] = 0

# populate from a list of records
records = [
    {"site":"Karachi", "month":"Jan", "tonnes":45.2},
    {"site":"Karachi", "month":"Feb", "tonnes":41.8},
    {"site":"Lahore", "month":"Jan", "tonnes":38.7},
    {"site":"Lahore", "month":"Feb", "tonnes":22.4},
    {"site":"Islamabad", "month":"Feb", "tonnes":51.9},
]
for r in records:
    totals[r["site"]][r["month"]] += r["tonnes"]

# print the grid - outer loop sites, inner loop months
print("SITE x MONTH GRID")
for site in sites:
    for month in months:
        v = totals[site][month]
        print(f" {site:<12} {month} -> {v:>5} tCO2e")
```
