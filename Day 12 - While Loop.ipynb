**Company:** A.P. Moller – Maersk 
** Maersk 2023 **Scope 1 GHG emissions = 34,138 thousand tCO₂e**.
**Source:** Maersk Sustainability Report 2023, page 20
  (https://www.maersk.com/sustainability, PDF publicly available).
**Modelling assumption ** annual total split evenly across 12 months
  → monthly = 34,138,000 ÷ 12 = **2,844,833 tCO₂e** per month.
- **Threshold used in Exercise 3:** 20,000,000 tCO₂e is our own alert level


```python
# The four part anatomy: initilise, test, work, update.
Monthly = 2844833
N_months = 12

# Version 1 - accumulate with a while loop
month = 0
total = 0
while month < N_months:
    total += Monthly
    print(f"After month {month+1}: total = {total:,} tCO2e")
    month += 1
print("--------------------------------")
print(f"year total from while loop: {total:,} tCO2e")

# Version 2 - Same thing without loop, using arithmetic
quick = Monthly * N_months
print(f"Year total from multiplication: {quick:,} tCO2e")

# Version 3 - Confirm both agree
print(f"Match: {total == quick}")
```

```python
# Break - stop as soon as a threshold is crossed.
# The threshold check pattern
monthly = 2844833
N_months = 12

# Version 1 - threshold will be breached mid-year
Threshold = 20000000
cummulative = 0
month = 0
while month < N_months:
    cummulative += Monthly
    print(f"Month {month+1}: cummulative = {cummulative:,} tCO2e")
    if cummulative > Threshold:
        print(f"Threshold breached in month {month+1}.")
        break
    month += 1
else:
    print("within threshold for the full year.")

print("----")

# Version 2 - Threshold is set so high it will never be breached
threshold_high = 999999999
cummulative = 0
month = 0
while month < N_months:
    cummulative += monthly
    if cummulative > threshold_high:
        print(f"Breached in month {month+1}")
        break
    month += 1
else:
    print(f"year finished at {cummulative:,} - within the threshold.")
```

```python
# Continue
# only include the verified records in total
records = [
    {"facility": "F001", "kg": 5200, "verified": True},
    {"facility": "F002", "kg": 4100, "verified": False},
    {"facility": "F003", "kg": 6300, "verified": True},
    {"facility": "F004", "kg": 2900, "verified": False},
]

# Version 1 - the correct order: advance the counter before continue
i = 0
verified_total = 0
while i < len(records):
    r = records[i]
    i += 1
    if not r["verified"]:
       print(f"skipping unverified {r['facility']}")
       continue
    verified_total += int(r["kg"])
    print(f"Added {r['facility']}: {r['kg']} kg")
print(f"verified total: {verified_total} kg")

print("----")
```

```python
# While True + Break - loop until a stop signal
# Simulate an interactive intake without actually calling input()
# In real use you would write: entry = input("kg (or 'done' / 'cancel'): ")

# Version 1 - one stop singal
feed = iter(["5200", "4100", "6300", "done", "999"])
total = 0
count = 0
while True:
    entry = next(feed)
    if entry == "done":
        break
    total += float(entry)
    count += 1
    print(f"Accepted {entry} kg (running total {total}")
print(f"stopped normally. entries: {count}, total: {total} kg")

print("----")

# Version 2 - two possible stop signals, distinguised after the loop
feed = iter(["5200", "4100", "cancel", "9999"])
total = 0
cancelled = False
while True:
    entry = next(feed)
    if entry == "done":
        break
    if entry == "cancel":
        cancelled = True
        break
    total += float(entry)
if cancelled:
    print(f"intake Cancelled after {total} kg - discarding.")
else:
    print(f"intake finished normally at {total} kg.")
```
