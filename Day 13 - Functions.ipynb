```python
# My first function - Def, Call, Return
# Version 1 - function that just prints (no return)
def greet_analyst():
    print("Hello, analyst! Ready to crunch some emissions?")
greet_analyst()

# Version 2 - a function that returns a value
def square(number):
    return number * number
result = square(9)
print(f"9 squared = {result}")

# Version 3 - What happens if you foget the parentheses?
print(square)
# version 3.1 
print(square(9))
```

```python
# Parameters vs Arguments
# Define a function with two parameters
def calc_emissions(quantity, factor):
    """Return the kg CO2e for an activity: quantity multiplied by an emission factor."""
    return quantity * factor

# Version 1 - correct order: quantity first, factor second
diesel_L = 12000
diesel_factor = 2.68
emissions = calc_emissions(diesel_L, diesel_factor)
print(f"Correct Call: {emissions:,.1f} kg CO2e")

# Version 2 - arguments swapped by mistake (python runs it anyway)
emission_wrong = calc_emissions(diesel_factor, diesel_L)
print(f"swapped call: {emission_wrong:,.1f} kg CO2e ")

# Version 3 - the safe way : argument labelling 
emissions_safe = calc_emissions(quantity=diesel_L, factor=diesel_factor)
print(f"Argument label: {emissions_safe:,.1f} kg CO2e ")

# Prove version 1 and version 3 give the same answer
print(f"Same Ans: {emission_wrong == emissions} ")
```

```python
# Building a real analytical tool - calc_emission() and scope_total() together 
# Dummy factors for learning
factors_kg_per_unit = {
    "diesel_L": 2.68,
    "gas_m3": 2.04,
    "petrol_L": 2.31,
}
# Function 1 - the atomic calculation
def calc_emissions(quantity, factor):
    """Return kg CO2e for one row: quantity * factor."""
    return quantity * factor

# Function 2 - reusese function 1 across the whole list
def scope_total(activities):
    """sum kg CO2e across a list of activity dicts."""
    total  = 0
    for a in activities: 
        factor = factors_kg_per_unit[a["type"]]
        total += calc_emissions(a["quantity"], factor)
    return total

# ------- Test the atomic function with 3 different fuel types---
print("per activity emissions")
print(f"Diesel 12000 L ->  {calc_emissions(12000, factors_kg_per_unit['diesel_L']):>12,.1f} kg CO2e ")
print(f"gas    40000 m3 -> {calc_emissions(40000, factors_kg_per_unit['gas_m3']):>12,.1f} kg CO2e ")
print(f"petrol 5500 L ->   {calc_emissions(5500, factors_kg_per_unit['petrol_L']):>12,.1f} kg CO2e ")

# Test scope_total with a small scope 1 activity ledger for one facility 
scope_1_activities = [
    {"facility": "F001", "type": "diesel_L", "quantity": 12000},
    {"facility": "F001", "type": "gas_m3", "quantity": 40000},
    {"facility": "F001", "type": "petrol_L", "quantity": 5500},
]

total = scope_total(scope_1_activities)
print(f"\nF001 Scope 1 total: {total:,.1f} kg CO2e")
```
