```python
# Positional vs Keyword Arguments
def cal_emissions(quantity, factor):
    return quantity * factor

# Positional -order must match def line
kg = cal_emissions(12400, 2.68)
print(f"Positional call: {kg:,.1f} kg CO2e")

# keyword argument - order can change, names make the call self documenting
kg = cal_emissions(factor=2.68, quantity=12400)
print(f"keyword argument: {kg:,.1f} kg CO2e")

# Mixed - Positional first, keyword after
kg = cal_emissions(12400, factor=2.68)
print(f"Mixec arugument: {kg:,.1f} kg CO2e")
```

```python
# Defualt Parameters - one function, many fuels
def calc_emissions(quantity, factor=2.68):
    """Return kg CO2e. Defaults to diesel factor (2.68 kg CO2e/L, illustrative)."""
    return quantity * factor

# use the defualt (diesel) - no factor needed
print(f"Assumed diesel: {calc_emissions(12400):,.1f} kg CO2e")

# Override for natural gas
print(f"Natural Gas: {calc_emissions(8200, factor=2.04):,.1f} kg CO2e ")

# Overide for grid electricity
print(f"Grid Electricity: {calc_emissions(45000, factor=0.42):,.1f} kg CO2e ")
```

```python
# Multiple return values - one function, once call, four stats
monthly_kg = [4200, 3900, 5100, 4800, 6300, 5200,
              5900, 6100, 4700, 4300, 3800, 4400]
def yearly_stats(values):
    """Return (total, mean, highest, lowest) for a list of monthly emissions."""
    total = sum(values)
    mean = total / len(values)
    highest = max(values)
    lowest = min(values)
    return total, mean, highest, lowest

# unpack all 4 into 4 named variables in one line
total, mean, highest, lowest = yearly_stats(monthly_kg)
print(f"Total: {total:,} kg CO2e")
print(f"Mean: {mean:,.1f} kg CO2e")
print(f"Highest: {highest:,} kg CO2 (bad month) ")
print(f"Lowest: {lowest:,} kg CO2e (Best month) ")
```

```python
# Local vs Global Scope - What function can and cannot see
default_factor = 2.68
def calc_diesel(litres):
    kg = litres * default_factor
    return kg

result = calc_diesel(12400)
print(f"Diesel emissions: {result:,.1f} kg CO2e")

# Try to reach 'litres' from outside the function. it does not exist here
try:
    print(litres)
except NameError as e:
    print(f"Name Error: {e}")
```

```python
# Assigning inside a function creates local
factor = 2.68
def wrong_update():
    factor = 3.00
    return factor
def right_update():
    global factor
    factor = 3.00
    return factor

print(f"Before wrong_update: global factor = {factor}")
inside = wrong_update()
print(f"inside wrong update: {inside} (local)")
print(f"After Wrong update: global factor = {factor} (unchanged) ")

inside = right_update()
print(f"After write update: global factor = {factor} (mutated) ")
```

```python
# Docstrings _ turning function into documentation
def calc_emissions(quantity, factor=2.68):
    """Calculate kg CO2e from an activity quantity and its emission factor.
 
    Args:
        quantity (float): The activity data (e.g. litres of diesel).
        factor (float, optional): kg CO2e per unit. Defaults to 2.68
            (diesel, illustrative teaching value).
 
    Returns:
        float: Emissions in kg CO2e.
 
    Example:
        >>> calc_emissions(1000)
        2680.0
    """
    return quantity * factor
 
# The docstring is now the function's official documentation.
# help() reads it, IDEs read it, and Jupyter's calc_emissions? reads it.
help(calc_emissions)
```

```python
# *args and **kwargs - aborbing any call shape 
# *args collects exrtra positional arguments into the tuple
def total_emissions(*amounts, factor=1.0):
    """Sum any number of facility amounts. all multiplied by one factor."""
    return sum(amounts) * factor

# Pass 4 diesel readings - they land in the 'amounts' tuple
print(f"{total_emissions(1200, 1450, 980, 1300, factor=2.68):,.1f} kg CO2e ")

# **kwargs collects extra keyword arguments into a dict
def facility_report(**metrics):
    """Print any key=value metrics passad as keyword arguments."""
    for label, value in metrics.items():
      print(f" {label:<12}: {value:,}")
    print(f" {'Total':<12}: {sum(metrics.values()):,}")

facility_report(diesel_kg=33232, gas_kg=16728, grid_kg=18900)
    
    
```

```python
# GHG_Summary 
def ghg_summary(emission_list, threshold=5000):
    """Compute ghg summary from emission list and comapre it with threshold value."""
    return {
    "total": sum(emission_list),
    "mean": round(sum(emission_list) / len(emission_list), 2),
    "max": max(emission_list),
    "min": min(emission_list),
    "count": len(emission_list),
    "count_above_threshold": sum(1 for v in emission_list if v > threshold),
    }

# Apply to a full year of monthly emisssions (kg CO2e)
monthly_kg = [4200, 3900, 5100, 4800, 6300, 5200, 5900, 6100, 4700, 4300, 3800, 4400]

summary = ghg_summary(monthly_kg, threshold=5000)
for label, value in summary.items():
    print(f"{label:<25}: {value:,}")
```
