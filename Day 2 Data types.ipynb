```python
# The four core data types in python
reporting_year = 2024
emission_factor = 2.68
activity = "Natural gas combustion"
is_verified = True

print(type(reporting_year))
print(type(emission_factor))
print(type(activity))
print(type(is_verified))
```

```python
# Convert between types using int(), float(), Str(), bool()
raw_quantity = "500"
print(type(raw_quantity))

# Convert text to a number so we can calculate
quantity = float(raw_quantity)
print(type(quantity))
print(quantity * 2.68)

# Convert a number back to text
year_int = 2024
year_str = str(year_int)
print("Reporting Year: " + year_str)

```

```python
activity = "Diesel generator combustion"
facility_id = 4
reporting_year = 2024
quantity_liters = 500.0
emission_factor_kg_per_1 = 2.68
is_verified = True

# Calculate emission
total_kg = quantity_liters * emission_factor_kg_per_1
total_tonnes = total_kg / 1000

# Report
print(f"Activity: {activity}")
print(f"facility: F00" + str(facility_id))
print(f"Year: {reporting_year}")
print(f"Total: {total_tonnes:.2f} tCO2e")
print(f"verified: {is_verified}")

print(type(activity), type(facility_id), type(quantity_liters), type(is_verified))
      
```
