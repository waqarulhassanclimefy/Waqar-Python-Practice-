```python
# The 7 artimatic operators
print(10 + 5)
print(10 - 5)
print(10 * 5)
print(10 / 3)
print(10 // 3)
print(10 % 3)
print(10 ** 3)

# Order of operations 
print(2 + 3 * 4)
print((2 + 3) * 4)
print(2 ** 3 + 1)

```

```python
# A client's total emissions in kg
total_kg = 2750
# How many full tonnes?
full_tonnes = total_kg // 1000
remainder_kg = total_kg % 1000

print(f"total: {total_kg} kg")
print(f"Full tonnes: {full_tonnes}")
print(f"Leftover kg: {remainder_kg}")

# Real fleet planning - How many full truck?
cargo_tonnes = 275
truck_capacity = 50
full_trucks = cargo_tonnes // truck_capacity
leftover_tonnes = cargo_tonnes % truck_capacity 

print(f"\nFull trucks needs: {full_trucks}")
print(f"partial truck load: {leftover_tonnes} tonnes")

```

```python
# A raw emission calculation
quanitity_liters = 507.3
emission_factor = 2.68345

total_kg = quanitity_liters * emission_factor
print(f"raw: {total_kg}")

# Round for display in a report
print(f"Rounded to 2 decimals: {round(total_kg, 2)}")

# Emission factor to 3 Decimals 
print(f"factor: {round(emission_factor, 3)}")

# Percentage share rounded to 1 decimals
scope1 = 145.6
total = 1516.9
share = (scope1 / total) * 100
print(f"scope 1 share: {round(share, 1)}%")


```

```python
# The infamous float precesion issue
a = 0.1 + 0.2
print(a)

# Are they equal?
print(a == 0.3)

# Two ways of safely compare
print(round(a, 5) == round(0.3, 5))

import math
print(math.isclose(a, 0.3))

# A real Climefy audit scenario
data1 = 0.1 + 0.2 + 0.3
data2 = 0.6

print(f"\ndata 1: {data1}")
print(f"Data 2: {data2}")
print(f"Naive Check: {data1 == data2}")
print(f"safe check: {math.isclose(data1, data2)}")


```

```python
import math

# sequre root - for chart scalling and standard deviation 
print(math.sqrt(144))
print(math.sqrt(2))

# Natural log for log scalling chart
print(math.log(1000))
print(math.log10(1000))
print(math.log10(1000000))

# Ceil - Always round up
cargo_tonnes = 275
truck_capacity = 50
truck_req = math.ceil(cargo_tonnes / truck_capacity)
print(f"truck required: {truck_req}")

# Floor - Always round Down
wheat_kg = 4750
per_shipment_kg = 1000
full_shipments = math.floor(wheat_kg / per_shipment_kg)
print(f"full shipments possible: {full_shipments}")

# The constant pi is also there
print(math.pi)
```

```python
import math

# one facility's fuel activity data
diesel_liters = 507.3
natural_gas_kwh = 12450
electricity_kwh = 8620

# Defra 2024 emission factors
diesel_factor = 2.68345
gas_factor = 0.18293
grid_factor = 0.21233

# Emission calculations
emissions_diesel_kg = diesel_liters * diesel_factor
emission_gas_kg = natural_gas_kwh * gas_factor
emission_grid_kg = electricity_kwh * grid_factor

total_kg = emissions_diesel_kg + emission_gas_kg + emission_grid_kg
total_tonnes = total_kg / 1000
# Rounded for the client report

print("=== Facility F004 - 2024 Emission Report ===")
print(f"diesel: {round(emissions_diesel_kg, 2)} kgcO2e")
print(f"Natural Gas: {round(emission_gas_kg, 2)} kgCO2e")
print(f"Electricity: {round(emission_grid_kg, 2)} kgCO2e")
print(f"Total: {round(total_kg, 2)} KGCO2e ({round(total_tonnes, 2)} tCO2e)")
```
