```python
# Way 1 - The literal syntax with { } curley braces
factors = {"diesel": 2.68, "natural gas": 2.02, "petrol": 2.31, "electricity": 0.42,}
print(factors)
print("number of fuels:", len(factors))

# Way 2 - dict() from a lisf of (key,value) tuples
pairs = [("diesel", 2.68), ("petrol", 2.31)]
d2 = dict(pairs)
print("From pairs:", d2)

# Way 3 - dict(zip(...)) from two parallel lists
fuels = ["diesel", "gas", "petrol"]
per_kg = [2.68, 2.02, 2.31]
d3 = dict(zip(fuels, per_kg))
print("From Zip:", d3)

           
```

```python
factors = {"diesel": 2.68, "gas": 2.02, "petrol": 2.31}
# Access with [] - fast when key exsists 
print("diesel factor:", factors["diesel"])

# Missing Key - Key error
# print(factors["coal"])

# .get() - safe access - return NONE if missing
print("coal factor:", factors.get("coal"))

# .get() with a defualt value - the recommended patter
print("coal factor:", factors.get("coal", 0.0))

# Real climefy pattern 
fuel = "LPG"
factor = factors.get(fuel, 0.0)
print(f"{fuel} factor: {factor}")

# check first. then use brackets
if "diesel" in factors:
    print("yes, diesel is:", factors["diesel"])

```

```python
factor = {"diesel": 2.68, "petrol": 2.31}

# add a new key - same syntax as update
factor["coal"] = 2.42
print("After add coal:", factor)

# Update existing key - Python treats it the same
factor["diesel"] = 2.70
print("After update:", factor)

# Delete with del - craches if the key missing
del factor["petrol"]
print("After del:", factor)

# .pop remove and returns the value
removed = factor.pop("coal")
print(f"popped coal ({removed}):", factor)

# .pop with default - safe when missing
missing = factor.pop("LPG", None)
print(f"popped missing: {missing}")

# .update() to merge another dict in
factor.update({"biomass": 0.39, "Solar": 0.05})
print("After merge:", factor)
```

```python
factors = {"diesel": 2.68, "gas": 2.02, "petrol": 2.31}

# in check the KEYS, not the values
print("diesel in?", "diesel" in factors) 
print("2.68 in?", 2.68 in factors)

# to check the values use .values()
print("2.68 in?:", 2.68 in factors.values())

# Loop over keys
print("---keys---")
for k in factors:
    print(" ", k)

# Loop over values
print("---Values---")
for v in factors.values():
    print(" ", v)

# Loop over both
print("---items---")
for x, y in factors.items():
    print(f" {x:<8} -> {y} kg/L")
```

```python
# Each facility is a dict - reads like a plain english
records = [
    {"id": "F001", "fuel":"diesel", "tonnes":45.2},
    {"id": "F002", "fuel":"gas", "tonnes":38.7},
    {"id": "F003", "fuel":"petrol", "tonnes":22.4},
    {"id": "F004", "fuel":"diesel", "tonnes":51.9},
]

# Access one field of one record
print("F002 tonnes:", records[1]["tonnes"])
print("F002 fuel:", records[1]["fuel"])

# Loop through records - clean f-string reports
for r in records:
    print(f"{r["id"]}: {r["tonnes"]} tCO2e ({r["fuel"]})")

# Filter - records above 30 tonnes
high = [r for r in records if r["tonnes"] > 30]
print("Higest emitter:", len(high))

# sort - by tonnes desending, using dict key access
ranked = sorted(records, key=lambda r: r["tonnes"], reverse=True)
print("Top emitted:", ranked[0]["id"])
```

```python
# Facility-level records
records = [
    {"id": "F001", "fuel":"diesel", "tonnes":45.2},
    {"id": "F002", "fuel":"gas", "tonnes":38.7},
    {"id": "F003", "fuel":"diesel", "tonnes":51.9},
    {"id": "F004", "fuel":"gas", "tonnes":29.7},
    {"id": "F005", "fuel":"petrol", "tonnes":22.4},
]

# Build dict of total by fuel using .get() 
totals = {}
for r in records:
    fuel = r["fuel"]
    totals[fuel] = totals.get(fuel, 0) + r["tonnes"]

print("Total by fuel:", totals)

# Also count how many facilities for fuel

counts = {}
for r in records:
    fuel = r["fuel"]
    counts[fuel] = counts.get(fuel, 0) +1 

print("counts by fuel:", counts)
# sort the total dict by value
ranked = sorted(totals.items(),
                key=lambda item: item[1], reverse=True)

# report using enumerate
print("ranked by fuel")
for i, (fuel, total) in enumerate(ranked, start=1):
    n = counts[fuel]
    print(f"{i}. {fuel:<8} - {round(total,2)} tCO2e ({n} sites)")
```
