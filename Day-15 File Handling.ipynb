```python
# (a) .read - Whole as a one string
with open("emissions_note.txt", "r", encoding="utf-8") as f:
    content = f.read()
print(content)
print("type:", type(content).__name__)

# (b) .readlines() - A list of line strings
with open("emissions_note.txt", "r", encoding="utf-8") as f:
    lines = f.readlines()
print(lines)
print("number of lines:", len(lines))

# (c) Iterate - one line at a time
with open("emissions_note.txt", "r", encoding="utf-8") as f:
    for line in f:
        print(repr(line))
```

# Clean and parse - .strip() then .split


```python
with open("emissions_note.txt", "r", encoding="utf-8") as f:
    lines = f.readlines()
data_lines = lines[1:]

for line in data_lines:
    clean = line.strip()
    activity, dt, kg = clean.split(",")
    print(f"{activity:<10} | {dt} | {float(kg):>8,.1f} kg")
```

```python
# Append mode "A" Add a line without overwriting
# "a" open at the end of the file and adds to it
with open("emissions_note.txt", "a", encoding="utf-8") as f:
    f.write("gas_m3,2025-01-14,1224.0\n")

# Read the whole file back to confirm the new line is there:
with open("emissions_note.txt", "r", encoding="utf-8") as f:
    print(f.read())
```

```python
# Try /Except - Servive a missing file
def read_log(path):
    try:
        with open(path, "r", encoding="utf-8") as f:
            return f.read()
    except FileNotFoundError:
        print(f"File not found: {path} - returning empty string.")
        return ""
good = read_log("emissions_note.txt")
print(f"Read {len(good)} characters from the real file.")

missing = read_log("does_not_exist.txt")
print(f"Missing File returned: {missing!r}")
```

```python
# The plan exercise - Build a 20 line log, total it, write a report
from datetime import date
#----- Part 1: Create emission_log.txt with 20 Lines

entries = [
    ("diesel_L","2025-01-06",3280.0), ("grid_kWh","2025-01-09",1890.5),
    ("gas_m3","2025-01-14",1224.0),   ("petrol_L","2025-01-20",742.3),
    ("diesel_L","2025-02-03",2960.0), ("grid_kWh","2025-02-11",2016.0),
    ("flight_pkm","2025-02-17",1564.0),("gas_m3","2025-02-25",1102.5),
    ("diesel_L","2025-03-04",3552.0), ("grid_kWh","2025-03-12",1743.0),
    ("petrol_L","2025-03-19",693.0),  ("gas_m3","2025-03-27",1346.4),
    ("diesel_L","2025-04-02",3016.0), ("grid_kWh","2025-04-15",2100.0),
    ("flight_pkm","2025-04-21",986.0),("petrol_L","2025-04-28",810.5),
    ("diesel_L","2025-05-06",3444.0), ("grid_kWh","2025-05-13",1932.0),
    ("gas_m3","2025-05-22",1428.0),   ("diesel_L","2025-05-29",3128.0),
]
with open("emission_log.txt", "w", encoding="utf-8") as f:
    for activity, dt, kg in entries:
        f.write(f"{activity},{dt},{kg}\n")
print(f"wrote emission_log.txt with {len(entries)} lines.")

# Read it back and total the kg column
total_kg = 0.0
count = 0
with open("emission_log.txt", "r", encoding="utf-8") as f:
    for line in f:
        activity, dt, kg = line.strip().split(",")
        total_kg += float(kg)
        count += 1
print(f"Read {count} activities. Total = {total_kg:,.1f} kg CO2e")

# Part 3: Write summary_report.txt with today's date
with open("summary_report.txt", "w", encoding="utf-8") as f:
    f.write("Climefy - Emission Summary Report\n")
    f.write(f"Generated: {date.today().isoformat()}\n")
    f.write(f"Activites counted: {count}\n")
    f.write(f"Total emissions:   {total_kg:,.1f} kg CO2e\n")
    f.write(f"                  = {total_kg/1000:,.3f} tonnes CO2e\n")

with open("summary_report.txt", "r", encoding="utf-8") as f:
 print(f.read())

```
