```python
# Diesel generator - Scope 1 emissions
activity = "Diesel generator combustion"
quantity_liters = 500
emission_factor_kg_per_liter = 2.68

total_kg = quantity_liters * emission_factor_kg_per_liter
total_tonnes = total_kg / 1000

print(f"Activity: {activity}")
print(f"Quantity: {quantity_liters} liters")
print(f"Emission factor: {emission_factor_kg_per_liter} kg/L")
print(f"Total: {total_kg} kg CO2e ({total_tonnes} tonnes CO2e)")
```
